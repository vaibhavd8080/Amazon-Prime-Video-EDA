````python
# Inspect the genres and production countries columns

display(titles[["title", "genres", "production_countries"]].head(10))
````

````python
# Import ast for safely converting list-like strings into Python lists

import ast
````

````python
# Convert genres and production countries from strings to Python lists

titles["genres"] = titles["genres"].apply(
    lambda x: ast.literal_eval(x) if pd.notna(x) else []
)

titles["production_countries"] = titles["production_countries"].apply(
    lambda x: ast.literal_eval(x) if pd.notna(x) else []
)
````

````python
# Verify that genres and production countries are now stored as lists

print("Genres example:", titles["genres"].iloc[0])
print("Genres data type:", type(titles["genres"].iloc[0]))

print("\nProduction countries example:", titles["production_countries"].iloc[0])
print("Production countries data type:", type(titles["production_countries"].iloc[0]))
````

````python
# Create a decade column from the release year

titles["release_decade"] = (titles["release_year"] // 10) * 10

# Check the result
titles[["release_year", "release_decade"]].head(10)
````

#### Chart - 8

````python
# Chart 8: Relationship between IMDb and TMDB ratings

rating_data = titles[["imdb_score", "tmdb_score"]].dropna()

plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=rating_data,
    x="imdb_score",
    y="tmdb_score",
    alpha=0.5
)

plt.title("IMDb Score vs TMDB Score")
plt.xlabel("IMDb Score")
plt.ylabel("TMDB Score")

plt.show()
````

##### 1. Why did you pick the specific chart?

A scatter plot is appropriate because both IMDb score and TMDB score are numerical variables. It helps us visually examine whether the two rating systems move together.


##### 2. What is/are the insight(s) found from the chart?

The scatter plot shows a positive relationship between IMDb and TMDB scores. Titles with higher IMDb scores generally tend to have higher TMDB scores, although the relationship is not perfect.

##### 3. Will the gained insights help creating a positive business impact?
Are there any insights that lead to negative growth? Justify with specific reason.

The positive relationship suggests that both rating platforms provide broadly consistent signals about audience reception. Stakeholders can consider ratings from both platforms when evaluating content performance, rather than relying on only one rating source.

#### Chart - 13

````python
# Chart 13: Content Type vs Age Certification

content_age = titles.copy()

# Treat missing age certifications as "Not Rated"
content_age["age_certification"] = content_age["age_certification"].fillna("Not Rated")

content_age_counts = pd.crosstab(
    content_age["age_certification"],
    content_age["type"]
)

plt.figure(figsize=(12, 6))

content_age_counts.plot(
    kind="bar",
    figsize=(12, 6)
)

plt.title("Content Type by Age Certification")
plt.xlabel("Age Certification")
plt.ylabel("Number of Titles")
plt.xticks(rotation=45)
plt.legend(title="Content Type")

plt.show()
````

##### 1. Why did you pick the specific chart?

A grouped bar chart is suitable for comparing two categorical variables: content type and age certification. It helps us understand whether the certification pattern differs between movies and TV shows.

##### 2. What is/are the insight(s) found from the chart?

Movies account for a larger number of titles across most age-certification categories because movies make up the majority of the catalog. The `Not Rated` category is also substantial, highlighting the large amount of missing certification information.

##### 3. Will the gained insights help creating a positive business impact?
Are there any insights that lead to negative growth? Justify with specific reason.

#### Chart - 14 - Correlation Heatmap

````python
# Chart 14: Correlation Heatmap of Numerical Variables

numeric_columns = [
    "release_year",
    "runtime",
    "seasons",
    "imdb_score",
    "imdb_votes",
    "tmdb_popularity",
    "tmdb_score"
]

correlation_matrix = titles[numeric_columns].corr()

plt.figure(figsize=(10, 8))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    linewidths=0.5
)

plt.title("Correlation Heatmap of Numerical Variables")

plt.show()
````

##### 1. Why did you pick the specific chart?

A correlation heatmap is useful for examining relationships among multiple numerical variables at the same time. The correlation coefficient ranges from -1 to +1, where values closer to +1 indicate a stronger positive relationship and values closer to -1 indicate a stronger negative relationship.


##### 2. What is/are the insight(s) found from the chart?

IMDb Score and TMDB Score show a moderate positive relationship, indicating that titles receiving higher ratings on IMDb generally tend to receive higher ratings on TMDB as well. The heatmap also helps identify whether popularity, votes, runtime, and seasons have strong or weak relationships with the rating variables.


**Business Impact:**  
Using multiple numerical indicators together gives stakeholders a broader view of content performance. Ratings, audience engagement, and popularity can be considered together when evaluating content rather than relying on a single metric.

The relationships between runtime and ratings appear less pronounced.

The analysis shows that movies form the majority of the catalog, while TV shows represent a smaller share. Drama and Comedy are among the most represented genres, and the United States is the largest production country, with India also making a significant contribution. The release-year analysis indicates a strong presence of newer content in the catalog.

The rating analysis shows that TV shows have higher average IMDb and TMDB ratings than movies in this dataset. IMDb and TMDB scores also show a positive relationship, indicating that the two rating platforms generally provide similar signals about audience reception. However, ratings and popularity should not be treated as the same measure, as highly voted or popular titles do not necessarily receive the highest ratings.

The analysis also identified missing values in important metadata fields such as age certification, ratings, votes, and popularity. These missing values should be handled carefully because some are expected by the nature of the data, while others may reduce the usefulness of the metadata.

Overall, the EDA provides a data-driven understanding of the Amazon Prime Video content catalog. These insights can help stakeholders make better decisions related to content acquisition, production, regional strategy, audience targeting, marketing, and content discovery. Future analysis could further explore viewer-level engagement, subscription behavior, and financial performance to make more precise business recommendations.
